## 공통 — 아래 모든 기능이 함께 쓰는 부분


````python
# 설정 — 앱 시작 시 1회. 모든 기능이 이 client 를 쓴다

import json, os, re
from pathlib import Path
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()   # 리포 루트 .env 의 GMS_KEY, GMS_BASE 를 읽는다

GMS_BASE = os.getenv("GMS_BASE", "https://gms.ssafy.io/gmsapi")

client = OpenAI(
    api_key=os.environ["GMS_KEY"],
    base_url=f"{GMS_BASE}/api.openai.com/v1",   # /v1 까지 필수. 빼면 404
)
print("base_url :", client.base_url)
````

````python
# 로그 읽기 — 공통. 나중에 이 부분만 Redis 조회로 바뀐다


LOG = Path("log_analysis/scenario/scenario1/발화로그-공개-STT.jsonl")

def load_utterances(path=LOG):
    with path.open(encoding="utf-8") as f:
        return [json.loads(line) for line in f if line.strip()]

ALL = load_utterances()
print(f"전체 {len(ALL)}건")
````

````python
# JSON 파싱 — 공통. 진술대조·칭호도 같은 방식으로 받는다

def strip_fence(text):
    """모델이 ```json 으로 감싸는 경우가 있어서 벗겨낸다."""
    return re.sub(r"^```(?:json)?\s*|\s*```$", "", text.strip())

````

# 1. 키워드 추출

````python
# 키워드 뽑기 — 발화 필터 · 프롬프트 입력 렌더

def for_keyword_extract(utterances, target, now):
    """지목 대상이 처음부터 지금까지 한 발화. 라운드를 제한하지 않는다.

    phase != NIGHT 는 안전장치다. 이 파일에는 밤 발화가 없지만,
    실수로 밤 로그를 합친 파일을 넣어도 여기서 걸린다.
    """
    return [u for u in utterances
            if u["speaker"] == target
            and u["t"] < now
            and u["phase"] != "NIGHT"]

def render(utterances):
    """u1 [R1] 발화내용 — 라운드를 붙이는 이유는 '반복'을 라운드별로 볼 수 있게 하려는 것"""
    return "\n".join(
        f"u{i} [R{u['round']}] {u['text']}"
        for i, u in enumerate(utterances, start=1)
    )

````

````python
# 키워드 뽑기 — 프롬프트

# 주의: 이 문자열에 .format() 을 쓰지 않는다. JSON 예시의 중괄호가 깨진다.
SYSTEM = """당신은 마피아 게임의 발언 기록을 분석하는 분석관이다.
지목된 플레이어가 게임 시작부터 지금까지 한 말에서, 반복적으로 등장한 키워드를 정확히 5개 뽑는다.

핵심 규칙 — 표현이 아니라 의미로 묶는다:
서로 글자가 완전히 다른 발언이라도 같은 태도나 같은 주장을 담고 있으면 하나로 묶어서 센다.
예를 들어 "시간 낭비다" / "삼분 동안 뭘 하겠나" / "방식이 비효율적이다" 는
글자가 다르지만 모두 토론 방식에 대한 불만이므로 하나의 키워드로 묶고 count 는 3이다.
글자가 같은 반복만 세면 5개를 채울 수 없다. 반드시 의미로 묶어라.

나머지 규칙:
1. 반드시 5개를 낸다. count 가 큰 것부터 순서대로 쓴다.
2. keyword 는 그 묶음을 대표하는 표현이고, 반드시 주어진 발언에 실제로 나온 말에서 고른다.
   없는 말을 새로 만들지 않는다. 10자 이내.
3. keyword 는 사전에 실릴 형태로 다듬어 쓴다. "-다" 로 끝내거나 명사로 끝낸다.
     원문 "억울하면"       → 억울하다
     원문 "근거도 없이"     → 근거 없다
     원문 "숨길 게 없어요"   → 숨길 게 없다
     원문 "시간 남비라고"    → 시간 낭비
   다만 다듬는 것은 어미까지다. 원문에 없는 단어를 새로 만들지 않는다.
   "소극적 태도", "근거 없는 주장" 처럼 발언을 요약해서 지은 이름은 쓰지 않는다.
4. 누구나 쓰는 말(그러니까, 저는, 네, 아니)은 고르지 않는다.
   그 사람의 태도나 입장이 드러나는 것을 고른다.
5. 누가 마피아인지 추측하지 않는다.
6. 아래 JSON 형식으로만 답한다. 설명이나 코드블록 표시를 붙이지 않는다.


{"keywords": [{"keyword": "시간 낭비", "count": 3, "evidence": ["u1", "u2", "u3"]}]}

count 는 그 의미가 등장한 발언 수.
evidence 는 근거가 된 발언 번호다. 반드시 주어진 번호 안에서만 쓴다."""


USER_TMPL = """분석 대상: {target}
현재 시점: {now}

발언 목록 (시간순):
{utterances}"""
````

````python
# 키워드 뽑기 — 호출

def extract_keywords(target, now, model="gpt-4o-mini", temperature=0, show_prompt=False):
    picked = for_keyword_extract(ALL, target, now)
    user = USER_TMPL.format(target=target, now=now, utterances=render(picked))

    if show_prompt:
        print("=== SYSTEM ===\n" + SYSTEM)
        print("\n=== USER ===\n" + user + "\n")

    resp = client.chat.completions.create(
        model=model,
        temperature=temperature,
        max_tokens=512,                  # gpt-4o-mini 는 max_tokens
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user",   "content": user},
        ],
    )
    raw = resp.choices[0].message.content
    try:
        parsed = json.loads(strip_fence(raw))
    except json.JSONDecodeError:
        parsed = None                    # 파싱 실패도 결과로 본다. 폴백 판단용
    return picked, raw, parsed
````

````python
# 키워드 뽑기 — 실행

TARGET = "딸기우유"
NOW    = "21:07:30"

_, raw, parsed = extract_keywords(TARGET, NOW)   # extract_keywords 함수가 호출 부분 포함

if parsed is None:
    print(raw)                      # JSON 파싱 실패 — 원문 그대로
else:
    for k in parsed.get("keywords", []):
        print(f"{k.get('keyword'):<12} count={k.get('count')}  {k.get('evidence')}")
````

# 2. 진술 대조

````python
# 진술대조권 — 발화 필터

def for_statement_compare(utterances, target, now):
    """지목 대상이 처음부터 지금까지 한 발화. 라운드를 제한하지 않는다.

    조건이 지금은 키워드 뽑기와 같다. 그래도 함수를 따로 두는 이유는
    한쪽 조건을 바꿀 때 다른 쪽이 딸려 바뀌지 않게 하려는 것이다.
    (키워드 뽑기는 원래 '현재 라운드만' 보는 설계였다)
    """
    return [u for u in utterances
            if u["speaker"] == target
            and u["t"] < now
            and u["phase"] != "NIGHT"]
````

````python
# 진술대조권 — 프롬프트

# 주의: 이 문자열에 .format() 을 쓰지 않는다. JSON 예시의 중괄호가 깨진다.
SYSTEM_SC = """당신은 마피아 게임의 발언 기록을 검토하는 분석관이다.
한 플레이어의 발언 목록을 받아서, 서로 앞뒤가 맞지 않는 발언을 찾아낸다.

발언은 "u1 [R1] 발언내용" 형태로 주어진다.
u1 은 발언 번호이고 [R1] 은 그 발언이 나온 라운드다. 라운드가 클수록 나중에 한 말이다.

규칙:
1. 주어진 발언 목록 밖의 내용은 절대 사용하지 않는다. 추측하거나 보충하지 않는다.
2. 발언 원문을 다시 쓰지 않는다. 발언 번호(u1, u2 ...)로만 지목한다.
3. 모순이 없으면 없다고 답한다. 이것은 실패가 아니라 정상적인 결과다.
   억지로 모순을 만들어내는 것이 이 작업에서 가장 나쁜 실패다.
4. 판단 기준은 "같은 대상에 대한 태도·사실 주장이 서로 양립할 수 없는가"다.
   생각이 바뀔 수 있는 의견 변화와, 과거를 다르게 진술하는 것은 다르다.
   후자만 모순으로 본다.
   모순이 아닌 예: "이제는 생각이 달라졌다", "마음이 바뀌었다",
                   새로운 근거가 나와서 의심 대상을 바꾼 것
   모순인 예: 과거에 의심한다고 말했는데 나중에 "처음부터 믿었다",
              "의심한 적이 없다" 고 말하는 것
5. 누가 마피아인지 추측하지 않는다. 발언 간의 일관성만 본다.
6. 짝을 짓기 전에 먼저 분류한다.
   대상 인물을 언급한 발언마다 그 인물에 대해 적대적인지 우호적인지 정한다.
   적대적: 수상하다, 의심한다, 투표하겠다, 처형에 찬성한다
   우호적: 믿는다, 의심한 적 없다, 아닌 것 같다, 감쌌다
   그 다음 적대적인 것과 우호적인 것 사이에서만 모순을 찾는다.
   적대끼리 또는 우호끼리 짝지은 것은 모순이 아니므로 절대 내지 않는다.
7. 발언의 일부가 음성인식 오류로 깨져 있어도, 뜻이 분명한 부분은 근거로 쓸 수 있다.
   예를 들어 "저는 A 씨가 제일 수상해요 ... 너를 생각입니다" 에서
   뒷부분은 깨졌지만 "제일 수상해요" 는 분명하므로 적대적 발언으로 센다.
   문장 전체가 무슨 말인지 알 수 없을 때만 제외한다.
8. 같은 발언 묶음을 두 번 내지 않는다.

형식:
9. summary 는 한 문장, 80자 이내. 무엇과 무엇이 충돌하는지만 쓴다.
10. reason 도 한 문장으로 쓴다.
11. utteranceIds 는 서로 충돌하는 발언 번호 2개 이상. 주어진 번호 안에서만 쓴다.
12. 아래 JSON 형식으로만 답한다. 설명이나 코드블록 표시를 붙이지 않는다.

모순을 찾았을 때:
{"hasFindings": true, "summary": "...", "contradictions": [{"utteranceIds": ["u2", "u6"], "reason": "..."}]}

모순이 없을 때:
{"hasFindings": false, "noFindingReason": "NO_CONTRADICTION", "contradictions": []}"""

USER_SC = """분석 대상: {target}
현재 시점: {now}

발언 목록 (시간순):
{utterances}"""
````

````python
# 진술대조권 — 호출
# 모델이 키워드 뽑기와 다르다. gpt-5.x 는 max_tokens 를 주면 HTTP 400 이므로
# max_completion_tokens 를 써야 한다.

def compare_statements(target, now, model="gpt-5.4-mini",
                       max_completion_tokens=2048, show_prompt=False):
    picked = for_statement_compare(ALL, target, now)

    # 임계값 — 3건 미만 또는 40자 미만이면 LLM 을 호출하지 않는다.
    # 호출하지 않으면 억지 모순이 만들어질 일도 없다.
    if len(picked) < 3 or sum(len(u["text"]) for u in picked) < 40:
        return picked, None, {"hasFindings": False,
                              "noFindingReason": "INSUFFICIENT",
                              "contradictions": []}

    user = USER_SC.format(target=target, now=now, utterances=render(picked))
    if show_prompt:
        print("=== SYSTEM ===\n" + SYSTEM_SC)
        print("\n=== USER ===\n" + user + "\n")

    resp = client.chat.completions.create(
        model=model,
        temperature=0,
        max_completion_tokens=max_completion_tokens,
        messages=[
            {"role": "system", "content": SYSTEM_SC},
            {"role": "user",   "content": user},
        ],
    )
    raw = resp.choices[0].message.content
    try:
        parsed = json.loads(strip_fence(raw))
    except json.JSONDecodeError:
        return picked, raw, None

    # 없는 번호는 버린다 — 이 아이템의 환각 방어. 프롬프트 품질과 무관하게 보장된다.
    valid = {f"u{i}" for i in range(1, len(picked) + 1)}
    for c in parsed.get("contradictions", []):
        ids = c.get("utteranceIds", [])
        bad = [i for i in ids if i not in valid]
        if bad:
            print("없는 번호 버림:", bad)
        c["utteranceIds"] = [i for i in ids if i in valid]

    return picked, raw, parsed
````

````python
# 진술대조권 — 실행

TARGET = "초코비"
NOW    = "21:11:05"     # 라면왕이 실제로 아이템을 쓴 시각

picked, raw, parsed = compare_statements(TARGET, NOW)

if parsed is None:
    print(raw)
elif not parsed.get("hasFindings"):
    print(parsed.get("noFindingReason"))
else:
    by_id = {f"u{i}": u for i, u in enumerate(picked, start=1)}
    print(parsed.get("summary"), "\n")
    for c in parsed.get("contradictions", []):
        print(c.get("utteranceIds"), c.get("reason"))
        for uid in c.get("utteranceIds", []):
            u = by_id[uid]
            print(f"   [R{u['round']} {u['t']}] {u['text']}")
        print()
````
